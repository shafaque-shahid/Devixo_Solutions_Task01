# **PART 1 – DATA LOADING & DATA UNDERSTANDING**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
import io

file_name = next(iter(uploaded))

df = pd.read_csv(io.BytesIO(uploaded[file_name]))

print("Dataset loaded successfully")
print("Dataset shape:", df.shape)
```

```python
print("First 5 records:")
display(df.head())

print("Last 5 records:")
display(df.tail())

print("Random sample of 5 records:")
display(df.sample(5, random_state=42))

print("Dataset information:")
df.info()

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset shape:", df.shape)
```

# **PART 2 – DATA CLEANING & PREPROCESSING**

```python
print("Original shape:", df.shape)

empty_rows = df.isna().all(axis=1).sum()

empty_columns = df.isna().all(axis=0).sum()

print("Completely empty rows:", empty_rows)
print("Completely empty columns:", empty_columns)

df = df.dropna(axis=0, how="all")
df = df.dropna(axis=1, how="all")

df.columns = df.columns.str.strip()

print("\nShape after removing empty rows and columns:", df.shape)
print("\nRemaining columns:")
print(df.columns.tolist())
```

```python
print("Total rows:", len(df))
print("Duplicate rows:", df.duplicated().sum())

print("\nMissing values in each column:")
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0].sort_values(ascending=False))

print("\nMissing value percentage:")
missing_percentage = (df.isnull().mean() * 100).round(2)
print(missing_percentage[missing_percentage > 0].sort_values(ascending=False))
```

```python
df_clean = df.copy()

df_clean["sales_commission_code"] = (
    df_clean["sales_commission_code"].fillna("Unknown")
)

df_category = df_clean.dropna(subset=["category_name_1"]).copy()

print("Original cleaned dataset:", df.shape)
print("Dataset after handling commission codes:", df_clean.shape)
print("Dataset for category analysis:", df_category.shape)

print("\nMissing values after handling commission codes:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python
print("Current data types:")
print(df_clean.dtypes)

print("\nSample values from important columns:")
for col in ["created_at", "price", "qty_ordered",
            "grand_total", "discount_amount", "Year",
            "Month", "Customer Since"]:
    print(f"\n{col}:")
    print(df_clean[col].dropna().head(5).tolist())
```

```python
df_clean["created_at"] = pd.to_datetime(
    df_clean["created_at"], errors="coerce"
)

df_clean["Working Date"] = pd.to_datetime(
    df_clean["Working Date"], errors="coerce"
)

df_clean["Customer Since"] = pd.to_datetime(
    df_clean["Customer Since"].astype("string"),
    format="%Y-%m",
    errors="coerce"
)

print("Updated date column types:")
print(df_clean[["created_at", "Working Date",
                "Customer Since"]].dtypes)

print("\nMissing dates after conversion:")
print(df_clean[["created_at", "Working Date",
                "Customer Since"]].isnull().sum())

print("\nDate range:")
print("First transaction:", df_clean["created_at"].min())
print("Last transaction:", df_clean["created_at"].max())
```

```python
# Final data quality check

print("Dataset shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])

print("\nNumeric summary:")
print(df_clean[[
    "price", "qty_ordered", "grand_total", "discount_amount"
]].describe())

print("\nTransaction status counts:")
print(df_clean["status"].value_counts(dropna=False).head(15))

print("\nPayment method counts:")
print(df_clean["payment_method"].value_counts(dropna=False))
```

```python
print("Rows with zero price:",
      (df_clean["price"] == 0).sum())

print("Rows with negative grand total:",
      (df_clean["grand_total"] < 0).sum())

print("Rows with negative discount:",
      (df_clean["discount_amount"] < 0).sum())

print("Rows with quantity above 100:",
      (df_clean["qty_ordered"] > 100).sum())

print("\nExamples of negative grand totals:")
print(df_clean.loc[
    df_clean["grand_total"] < 0,
    ["status", "price", "qty_ordered",
     "grand_total", "discount_amount", "payment_method"]
].head(10))

print("\nExamples of quantities above 100:")
print(df_clean.loc[
    df_clean["qty_ordered"] > 100,
    ["status", "sku", "price", "qty_ordered", "grand_total"]
].head(10))
```

```python
quality_report = pd.DataFrame({
    "Issue": [
        "Completely empty rows removed",
        "Completely empty columns removed",
        "Exact duplicate rows",
        "Missing commission codes filled",
        "Missing categories",
        "Zero-price transactions",
        "Negative grand totals",
        "Negative discounts",
        "Quantities above 100"
    ],
    "Count": [
        464051,
        5,
        df.duplicated().sum(),
        137178,
        df_clean["category_name_1"].isna().sum(),
        (df_clean["price"] == 0).sum(),
        (df_clean["grand_total"] < 0).sum(),
        (df_clean["discount_amount"] < 0).sum(),
        (df_clean["qty_ordered"] > 100).sum()
    ]
})

print(quality_report.to_string(index=False))

print("\nFinal dataset shape:", df_clean.shape)
print("Transaction date range:",
      df_clean["created_at"].min().date(), "to",
      df_clean["created_at"].max().date())
```

# **PART 3 – EXPLORATORY DATA ANALYSIS (EDA)**

```python
print("Numerical Summary")
print(df_clean[[
    "price", "qty_ordered", "grand_total", "discount_amount"
]].describe().round(2))

print("\n Number of Unique Values")
for col in ["status", "category_name_1",
            "payment_method", "Customer ID"]:
    print(f"{col}: {df_clean[col].nunique()} unique values")

print("\nTop 10 Product Catogories")
print(df_clean["category_name_1"].value_counts().head(10))

print("\nTop 10 Payment Methods")
print(df_clean["payment_method"].value_counts().head(10))

print("\nTop 10 transaction status")
print(df_clean["status"].value_counts().head(10))
```

```python
df_clean["YearMonth"] = df_clean["created_at"].dt.to_period("M")

monthly_records = (
    df_clean.groupby("YearMonth")
    .size()
    .sort_index()
)

print("Monthly Trasnsaction Records")
print(monthly_records.to_string())

print("\nMonth with the highest number of records:")
print(monthly_records.idxmax(), "-", monthly_records.max(), "records")

print("\nMonth with the lowest number of records:")
print(monthly_records.idxmin(), "-", monthly_records.min(), "records")
```

```python
category_counts = df_clean["category_name_1"].value_counts()
category_percentages = (
    category_counts / df_clean["category_name_1"].notna().sum() * 100
).round(2)

category_summary = pd.DataFrame({
    "Transaction Records": category_counts,
    "Percentage (%)": category_percentages
})

print("Top 10 Categories")
print(category_summary.head(10).to_string())

print("\nTotal distinct categories:", df_clean["category_name_1"].nunique())
print("Records with known category:",
      df_clean["category_name_1"].notna().sum())
```

```python
status_counts = df_clean["status"].value_counts()

status_percentages = (
    df_clean["status"].value_counts(normalize=True) * 100
).round(2)

status_summary = pd.DataFrame({
    "Transaction Records": status_counts,
    "Percentage (%)": status_percentages
})

print("Transaction Status Summary")
print(status_summary.to_string())

print("\nTotal transaction records:", len(df_clean))
print("Records with a known status:",
      df_clean["status"].notna().sum())
```

```python
payment_counts = df_clean["payment_method"].value_counts()

payment_percentages = (
    df_clean["payment_method"].value_counts(normalize=True) * 100
).round(2)

payment_summary = pd.DataFrame({
    "Transaction Records": payment_counts,
    "Percentage (%)": payment_percentages
})

print("Payment Method Summary")
print(payment_summary.to_string())

print("\nNumber of distinct payment methods:",
      df_clean["payment_method"].nunique())
```

```python
category_price_summary = (
    df_clean.dropna(subset=["category_name_1"])
    .groupby("category_name_1")
    .agg(
        Transaction_Records=("price", "size"),
        Average_Price=("price", "mean"),
        Median_Price=("price", "median"),
        Average_Discount=("discount_amount", "mean")
    )
    .sort_values("Transaction_Records", ascending=False)
    .round(2)
)

print("Price and Discount by Category")
print(category_price_summary.to_string())

print("\nCategory with highest median price:")
print(category_price_summary["Median_Price"].idxmax())

print("\nCategory with lowest median price:")
print(category_price_summary["Median_Price"].idxmin())
```

```python
print("EDA Summary")

print("Total transaction records:", len(df_clean))

print("Unique customers:",
      df_clean["Customer ID"].nunique())

print("Unique product categories:",
      df_clean.loc[
          ~df_clean["category_name_1"].isin([r"\N"]),
          "category_name_1"
      ].nunique())

print("Average item price:",
      round(df_clean["price"].mean(), 2))

print("Median item price:",
      round(df_clean["price"].median(), 2))

print("Average quantity per record:",
      round(df_clean["qty_ordered"].mean(), 2))

print("Most common category:",
      df_clean["category_name_1"].replace(r"\N", np.nan)
      .value_counts().idxmax())

print("Most common payment method:",
      df_clean["payment_method"].value_counts().idxmax())

print("Most common transaction status:",
      df_clean["status"].value_counts().idxmax())

print("Peak month by record count:",
      monthly_records.idxmax(),
      "-", monthly_records.max(), "records")
```

# **PART 4 – Data Visualization**

```python
# 1: Histogram

plt.figure(figsize=(10, 5))

sns.histplot(data=df_clean, x="price", bins=50)

plt.xscale("log")
plt.title("Distribution of Item Prices")
plt.xlabel("Item Price (log scale)")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.show()
```

```python
# 2: Top 10 Product Categories

category_counts = (
    df_clean[
        ~df_clean["category_name_1"].isin([r"\N"])
    ]["category_name_1"]
    .dropna()
    .value_counts()
    .head(10)
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=category_counts.values,
    y=category_counts.index
)

plt.title("Top 10 Product Categories by Transaction Records")
plt.xlabel("Number of Records")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()
```

```python
# 3: Payment Method Distribution

payment_counts = df_clean["payment_method"].value_counts()

top_payments = payment_counts.head(5)
other_count = payment_counts.iloc[5:].sum()

pie_data = pd.concat([
    top_payments,
    pd.Series({"Other": other_count})
])

plt.figure(figsize=(9, 7))

plt.pie(
    pie_data,
    labels=pie_data.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Payment Methods")
plt.tight_layout()
plt.show()
```

```python
# 4: Price vs Quantity

sample_data = df_clean.sample(
    n=min(5000, len(df_clean)),
    random_state=42
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=sample_data,
    x="price",
    y="qty_ordered",
    alpha=0.4
)

plt.xscale("log")
plt.title("Relationship Between Item Price and Quantity Ordered")
plt.xlabel("Item Price (log scale)")
plt.ylabel("Quantity Ordered")
plt.tight_layout()
plt.show()
```

```python
# 5: Monthly Transaction Trends

monthly_records = (
    df_clean.groupby(df_clean["created_at"].dt.to_period("M"))
    .size()
    .sort_index()
)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_records.index.astype(str),
    monthly_records.values,
    marker="o"
)

plt.title("Monthly Transaction Record Trends")
plt.xlabel("Month")
plt.ylabel("Number of Records")
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# 6: Price Distribution by Category

plot_data = df_clean[
    df_clean["category_name_1"].notna()
    & (df_clean["category_name_1"] != r"\N")
    & (df_clean["price"] > 0)
].copy()

top_categories = (
    plot_data["category_name_1"]
    .value_counts()
    .head(8)
    .index
)

plot_data = plot_data[
    plot_data["category_name_1"].isin(top_categories)
]

plt.figure(figsize=(13, 7))

sns.boxplot(
    data=plot_data,
    x="category_name_1",
    y="price"
)

plt.yscale("log")
plt.title("Item Price Distribution Across Top 8 Categories")
plt.xlabel("Product Category")
plt.ylabel("Item Price (log scale)")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

```python
# 7: Correlation Heatmap

numeric_cols = [
    "price",
    "qty_ordered",
    "grand_total",
    "discount_amount",
    "Year",
    "Month"
]

correlation_matrix = df_clean[numeric_cols].corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()
```

```python
# 8: Transaction Status Counts

status_data = df_clean.copy()

# Treat the dataset's \N marker as missing
status_data["status"] = status_data["status"].replace(r"\N", np.nan)

plt.figure(figsize=(12, 6))

sns.countplot(
    data=status_data,
    x="status",
    order=status_data["status"].value_counts().index
)

plt.title("Distribution of Transaction Statuses")
plt.xlabel("Transaction Status")
plt.ylabel("Number of Records")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# 9: Pair Plot

pair_data = df_clean[
    ["price", "qty_ordered", "grand_total", "discount_amount"]
].dropna()

pair_data = pair_data.sample(
    n=min(1000, len(pair_data)),
    random_state=42
)

sns.pairplot(
    pair_data,
    diag_kind="hist",
    corner=True
)

plt.suptitle(
    "Pair Plot of Numerical Variables",
    y=1.02
)

plt.show()
```

```python
# 10: Violin Plot

violin_data = df_clean[
    df_clean["category_name_1"].notna()
    & (df_clean["category_name_1"] != r"\N")
    & (df_clean["price"] > 0)
].copy()

top_categories = (
    violin_data["category_name_1"]
    .value_counts()
    .head(8)
    .index
)

violin_data = violin_data[
    violin_data["category_name_1"].isin(top_categories)
].sample(
    n=min(5000, len(violin_data)),
    random_state=42
)

plt.figure(figsize=(13, 7))

sns.violinplot(
    data=violin_data,
    x="category_name_1",
    y="price",
    cut=0
)

plt.yscale("log")
plt.title("Item Price Distribution by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Item Price (log scale)")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```
